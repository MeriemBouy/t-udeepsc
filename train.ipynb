```python
import sys
# --- JUPYTER FIX ---
sys.argv = [''] 

import torch
from transformers import BertTokenizer
from base_args import get_args
from utils import get_model

# --- UDEEPSC MODEL FIX ---
import model 

def test_single_description(text, checkpoint_path):
    # 1. Setup device and tokenizer
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

    # 2. Fake the arguments needed to build the model
    args = get_args()
    args.model = 'UDeepSC_new_model'
    args.ta_perform = 'textc'
    args.nb_classes = 8
    args.device = 'cuda' if torch.cuda.is_available() else 'cpu'
    
    # 3. Build model and load your trained weights
    print(f"Creating and loading model from {checkpoint_path}...")
    model_instance = get_model(args)
    
    # --- PYTORCH 2.6 FIX ---
    checkpoint = torch.load(checkpoint_path, map_location=device, weights_only=False)
    
    # UDeepSC usually saves the weights under the 'model' key
    if 'model' in checkpoint:
        model_instance.load_state_dict(checkpoint['model'], strict=False)
    else:
        model_instance.load_state_dict(checkpoint, strict=False)
        
    model_instance.to(device)
    model_instance.eval() # Set to evaluation mode

    # 4. Tokenize your custom text
    encoding = tokenizer(
        text, 
        max_length=128, 
        padding='max_length', 
        truncation=True, 
        return_tensors='pt'
    )
    
    input_ids = encoding['input_ids'].to(device)

    # 5. Get the prediction
    with torch.no_grad():
        # ---> THE FINAL FIX IS HERE <---
        # We explicitly tell the model we are doing text classification with 0 noise
        outputs = model_instance(text=input_ids, ta_perform='textc')
        
        predicted_idx = torch.argmax(outputs, dim=1).item()

    # 6. Map the integer back to the string class
    label_map = {
        0: "AccidentAhead", 
        1: "ClearRoad", 
        2: "LaneClosure", 
        3: "LowVisibility", 
        4: "PriorityVehicle", 
        5: "SlipperyRoad",
        6: "TrafficJam",
        7: "WorkZone"
    }
    
    predicted_class = label_map.get(predicted_idx, f"Unknown Index: {predicted_idx}")
    
    print("\n" + "="*70)
    print(f"INPUT TEXT: '{text}'")
    print(f"PREDICTION: {predicted_class} (Index {predicted_idx})")
    print("="*70 + "\n")


# --- RUN THE TEST ---

my_checkpoint = r"ckpt_record\ckpt_textc\checkpoint-199.pth" 
my_text = "Rsu camera array reports normal traffic flow on the suburban boulevard southbound at 14:58. Visibility is excellent."

test_single_description(my_text, my_checkpoint)
```

```python
# New dataset
import os

def separate_cases(input_filename, valid_filename, other_filename):
    # 1. Check if the file actually exists
    if not os.path.exists(input_filename):
        print(f"ERROR: Could not find '{input_filename}'. Make sure it's in the same folder as this script.")
        return

    # 2. Read the file, ignoring encoding errors
    with open(input_filename, 'r', encoding='utf-8', errors='ignore') as f:
        lines = f.readlines()
        
    print(f"Successfully read {len(lines)} lines from '{input_filename}'.")

    # 3. Check if the original file is actually empty
    if len(lines) == 0:
        print("ERROR: The input file is completely empty! Please check your original UDeepSC_naturalized_dataset.txt file to ensure it has the data.")
        return

    valid_count = 0
    other_count = 0

    with open(valid_filename, 'w', encoding='utf-8') as f_valid, \
         open(other_filename, 'w', encoding='utf-8') as f_other:
         
        current_block = []
        in_case_block = False
        is_valid = False

        for line in lines:
            stripped_line = line.strip()
            
            # Detect the start of a case (ignores leading/trailing spaces)
            if stripped_line.startswith("CLASS:"):
                in_case_block = True
            
            if in_case_block:
                current_block.append(line)
                
                # Check the TYPE within the block
                if stripped_line.startswith("TYPE: VALID"):
                    is_valid = True
                
                # Detect the end of a case block
                if stripped_line == "---":
                    if is_valid:
                        f_valid.writelines(current_block)
                        valid_count += 1
                    else:
                        f_other.writelines(current_block)
                        other_count += 1
                    
                    # Reset variables for the next block
                    current_block = []
                    in_case_block = False
                    is_valid = False
            else:
                # Write global headers/footers to both files
                f_valid.write(line)
                f_other.write(line)
                
        # Catch any trailing block that doesn't end in '---'
        if current_block:
            if is_valid:
                f_valid.writelines(current_block)
                valid_count += 1
            else:
                f_other.writelines(current_block)
                other_count += 1

    print(f"Done! Saved {valid_count} VALID cases into '{valid_filename}'.")
    print(f"Saved {other_count} OTHER cases into '{other_filename}'.")

# Run the separation process
separate_cases('ADS_dataset.txt', 'valid_cases.txt', 'other_cases.txt')
```

```python
import os

def format_dataset(input_file, output_file):
    # 1. Verify the original file exists
    if not os.path.exists(input_file):
        print(f"ERROR: Could not find '{input_file}'. Make sure it is in the same folder as this script.")
        return

    # 2. Read the file, safely ignoring encoding errors
    with open(input_file, 'r', encoding='utf-8', errors='ignore') as f:
        lines = f.readlines()

    valid_count = 0
    
    # 3. Process and write directly to the new format
    with open(output_file, 'w', encoding='utf-8') as f_out:
        current_class = ""
        current_text = []
        is_valid = False
        in_text_block = False

        for line in lines:
            stripped = line.strip()

            # Identify the class name
            if stripped.startswith("CLASS:"):
                current_class = stripped.replace("CLASS:", "").strip()
            
            # Check if this case is VALID
            elif stripped.startswith("TYPE:"):
                if "VALID" in stripped:
                    is_valid = True
            
            # Start capturing the description text
            elif stripped.startswith("TEXT:"):
                in_text_block = True
            
            # The '---' marks the end of a block
            elif stripped == "---":
                # Only write to file if it was marked as VALID
                if is_valid and current_class and current_text:
                    # Clean up the text block
                    text_joined = "".join(current_text).strip()
                    
                    # Write in the exact requested format
                    f_out.write("Description:\n")
                    f_out.write(text_joined + "\n")
                    f_out.write("SemanticClass:\n")
                    f_out.write(current_class + "\n\n")
                    
                    valid_count += 1
                
                # Reset variables for the next case block
                current_class = ""
                current_text = []
                is_valid = False
                in_text_block = False
            
            # If we are inside the text block, keep appending the lines
            elif in_text_block:
                if stripped: # Skip adding pure blank lines inside the description
                    current_text.append(line)

    print(f"Success! Formatted {valid_count} VALID cases and saved them to '{output_file}'.")

# Run the formatting process
format_dataset('valid_cases.txt', 'formatted_valid_cases.txt')
```
